Задание 6. Анализ жанров — количество игр

Задание 6: считаем количество игр каждого жанра и сортируем

```python
genre_count = df_clean.groupby('Genre')['Name'].count()
genre_count.sort_values(ascending=False)
```

Задание 7. Продажи по жанрам

Задание 7: суммарные мировые продажи по жанрам, отсортированные

```python
genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum()
genre_sales.sort_values(ascending=False)
```

Задание 8. Самые продаваемые игры

Задание 8: топ-10 игр по мировым продажам, только нужные столбцы

```python
df_clean.nlargest(10, 'Global_Sales')[['Rank', 'Name', 'Platform', 'Genre', 'Global_Sales']]
```

Задание 9. Анализ платформ

Задание 9: количество игр для каждой платформы, топ-10

```python
platform_count = df_clean.groupby('Platform')['Name'].count()
platform_count.sort_values(ascending=False).head(10)
```

Задание 9: платформы с наибольшими суммарными продажами

```python
platform_sales = df_clean.groupby('Platform')['Global_Sales'].sum()
platform_sales.sort_values(ascending=False).head(10)
```

Задание 10. Визуализация продаж по жанрам

Задание 10: столбчатая диаграмма суммарных продаж по жанрам

```python
genre_sales.sort_values(ascending=False).plot(kind='bar', figsize=(10, 5))
plt.title('Мировые продажи по жанрам')
plt.xlabel('Жанр')
plt.ylabel('Продажи, млн')
plt.tight_layout()
plt.show()
```

Наибольшие продажи у жанра Action

Наименьшие — у Strategy

Разница между топовым и последним жанром — более чем в 30 раз

Задание 12. Диаграмма рассеяния

Задание 12: диаграмма рассеяния NA_Sales и Global_Sales

```python
plt.figure(figsize=(8, 5))
plt.scatter(df_clean['NA_Sales'], df_clean['Global_Sales'], alpha=0.5)
plt.title('Связь продаж в Северной Америке и мировых продаж')
plt.xlabel('Продажи в Северной Америке, млн')
plt.ylabel('Мировые продажи, млн')
plt.show()
```

Одна точка — одна игра

Восходящая тенденция наблюдается

График соответствует результату корреляционного анализа
